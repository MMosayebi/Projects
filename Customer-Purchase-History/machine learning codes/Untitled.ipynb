```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import openpyxl
# pip install pycaret
```

```python
data = pd.read_excel("../dataset/Customer-Purchase-History.xlsx")
data.head(4)
```

```python
sns.boxplot(data=data, x="TotalPrice")
plt.show()
```

```python
print(f"Old data shape: {data.shape}")

Q1 = data["TotalPrice"].quantile(0.25)
Q3 = data["TotalPrice"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = data.loc[(data["TotalPrice"] >= lower) & (data["TotalPrice"] <=upper)]

print(f"New data shape: {df.shape}")
```
