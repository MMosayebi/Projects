```python
import pandas as pd
import numpy as np
from pycaret.regression import *
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
df = pd.read_excel("data transformed/data.xlsx")
```

```python
df.head(3)
```

```python
X = df[["Area","Room","Parking","Warehouse","Elevator"]]
y = df["Price"]


new_df = X.copy()
new_df["Price"] = y
```

```python
reg = setup(data=new_df,target="Price",session_id=42,train_size=0.7,normalize=True)
```

```python
best_model = compare_models()
```

```python
et_model = create_model("et")
```

```python
train_predictions = predict_model(et_model,data=get_config("X_train"))
y_train = get_config("y_train")
y_train_pred = train_predictions["prediction_label"]


test_predictions = predict_model(et_model)
y_test = test_predictions["Price"]
y_test_pred = test_predictions["prediction_label"]
```

```python
r2_score_train = r2_score(y_train, y_train_pred)
r2_score_test = r2_score(y_test, y_test_pred)

r2_gap = r2_score_train - r2_score_test

mae = mean_absolute_error(y_test, y_test_pred)
mse = mean_squared_error(y_test, y_test_pred)
rmse = np.sqrt(mse)
```

```python
print(f"r2_score of train data: {r2_score_train:,.2f}")
print(f"r2_score of test data: {r2_score_test:,.2f}")
print(f"Gap between train and test: {r2_score_train - r2_score_test:,.2f}\n")

print(f"MAE: {mae:,.0f}")
print(f"MSE: {mse:,.0f}")
print(f"RMSE: {rmse:,.0f}")
```

```python
final_et_model = finalize_model(et_model)
```

```python
unseen_data = pd.DataFrame({"Area":[50,45,70],
                           "Room":[2,1,2],
                           "Parking":[0,1,1],
                           "Warehouse":[1,1,1],
                           "Elevator":[0,1,1]})

predictions = predict_model(
    final_et_model,
    data=unseen_data
)

predictions
```
